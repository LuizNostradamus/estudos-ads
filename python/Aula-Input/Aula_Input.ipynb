```python
### ATIVIDADE 03 ###
## >int< use para numeros inteiros, >float< para numeros quebrados ##

valor1 = float(input("Digite o seu valor:"))
valor2 = float(input("Digite o seu valor:"))
soma = valor1 + valor2
subtracao = valor1 - valor2
divisao = valor1 / valor2
multi = valor1 * valor2
print("Soma do valor:", soma, "\nSubtração do valor:", subtracao, "\nDivisão do valor:", divisao, "\nMutiplicação do valor:", multi)
```
