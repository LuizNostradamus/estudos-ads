METADOS DE STRING


```python
### lower() ###
name = "LUIZ"
print(name.lower())
```

```python
### upper() ###
name = "luiz"
print(name.upper())
```

```python
### strip() ###
name = " Luiz "
print(name.strip())
```

```python
### replace() ###
text = "Luiz"
print(text.replace("Luiz", "Lewis"))
```

```python
### len() ###
text = "Luiz"
print(len(text))
```

```python
### ATIVIDADE 01 ###
name = " MARIA EDUARDA ALMEIDA "
email = " MARIA.ALMEIDA@EMPRESA.COM "

# 1 - REMOVER ESPAÇOS EXTRAS #
name = name.strip()
email = email.strip()
# 2 - DEIXAR MINUSCULO #
name = name.lower()
email = email.lower()
# 3 - SUBSTITUIR O DOMINIO DO EMAIL #
email = email.replace("empresa.com", "gmail.com")
# 4 - QUANTIDADES DE CARACTERES DO NOME #
quant = len(name)
# 5 -  EXIBA A ULTIMA/PRIMEIRA LETRA DO NOME #
ult_letra = name[-1]
pri_letra = name[0]

# name = input("Nome do usario: ")
print("Nome do usuario:", name)
print("Email do usuario:", email)
print("Utima letra do nome:", ult_letra)
print("Primeira letra do nome:", pri_letra)
print("Quantidade de letras do nome:", quant)

```
